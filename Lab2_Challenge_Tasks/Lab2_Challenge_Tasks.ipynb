```python
# [CELL 1] CHALLENGE CONFIGURATION
ENGINEER = "Arcilla"
STUDENT_ID = "TUPM-26-1087"

# Derive the Logic Seed from the last digit of the Student ID
SEED_NUM = int(STUDENT_ID[-1])

print(f"System Initialized for: {ENGINEER}")
print(f"Algorithmic Seed Value: {SEED_NUM}")
```

```python
# [CELL 2] SENSOR DATA GENERATION

sensor_data = []

for letter in ENGINEER:
    value = ord(letter) + SEED_NUM
    sensor_data.append(value)

# Fixed number of sensor readings
sensor_data = sensor_data[:5]

print("Generated Sensor Data:", sensor_data)
```

```python
# [CELL 3] OPERATING RANGE

limit = (SEED_NUM * 10) + 20

print("Safety Limit:", limit)
```

```python
# [CELL 4] RECEIVE AND VALIDATE READINGS

valid_readings = []
invalid_readings = []
classification_results = []

for i in range(5):

    user_input = input("Enter Sensor Reading " + str(i + 1) + ": ")

    try:
        reading = int(user_input)

        # Validate the reading
        if reading < 0:
            print("Invalid Reading")
            invalid_readings.append(user_input)

        else:
            print("Valid Reading")
            valid_readings.append(reading)

            # Classify valid readings
            if reading < 5:
                result = "LOW SIGNAL"

            elif reading <= limit:
                result = "NOMINAL"

            else:
                result = "OVER LIMIT"

            print("Classification:", result)
            classification_results.append(result)

    except ValueError:
        print("Invalid Input: Please enter a number.")
        invalid_readings.append(user_input)
```

```python
# [CELL 5] FINAL SUMMARY

print("------ FINAL OUTPUT ------")

print("Generated Sensor Data:", sensor_data)
print("Valid Results:", valid_readings)
print("Invalid Results:", invalid_readings)
print("Classification Results:", classification_results)

print("Total Readings:", 5)
print("Total Valid:", len(valid_readings))
print("Total Invalid:", len(invalid_readings))
print("Safety Limit:", limit)
```

```python
# [CELL 6] EXERCISE 2 CONFIGURATION

ENGINEER = "Arcilla"
STUDENT_ID = "TUPM-26-1087"
SEED_NUM = int(STUDENT_ID[-1])

print("System Initialized for:", ENGINEER)
print("Seed Number:", SEED_NUM)
```

```python
# [CELL 7] GENERATE SIGNAL

signal = ENGINEER + str(SEED_NUM)

print("Generated Signal:", signal)
```

```python
# [CELL 8] PROCESS AND NORMALIZE SIGNAL

processed_signal = signal.upper()

print("Processed Signal:", processed_signal)
```

```python
# [CELL 9] CHARACTER ANALYSIS

letters = 0
numbers = 0
spaces = 0
others = 0

print("------ CHARACTER ANALYSIS ------")

for character in processed_signal:

    if character.isalpha():
        print(character, "- LETTER")
        letters += 1

    elif character.isdigit():
        print(character, "- NUMBER")
        numbers += 1

    elif character.isspace():
        print(character, "- SPACE")
        spaces += 1

    else:
        print(character, "- OTHER")
        others += 1
```

```python
# [CELL 10] SIGNAL CLASSIFICATION

if others > 0:
    classification = "CONTAINS OTHER CHARACTERS"

elif numbers > 0:
    classification = "ALPHANUMERIC SIGNAL"

else:
    classification = "LETTER SIGNAL"

print("Signal Classification:", classification)
```

```python
# [CELL 11] DIAGNOSTIC REPORT

print("===== DIAGNOSTIC REPORT =====")
print("Generated Signal:", signal)
print("Processed Signal:", processed_signal)

print("Character Analysis:")
print("Letters:", letters)
print("Numbers:", numbers)
print("Spaces:", spaces)
print("Other Characters:", others)

print("Signal Classification:", classification)
```

```python
# [CELL 12] ACCESS SYSTEM CONFIGURATION

ENGINEER = "Arcilla"
STUDENT_ID = "TUPM-26-1087"
SEED_NUM = int(STUDENT_ID[-1])

print("=== ACCESS SYSTEM INITIALIZED ===")
print("Engineer:", ENGINEER)
print("Seed Number:", SEED_NUM)
```

```python
# [CELL 13] ACCESS CREDENTIAL

access_key = (ENGINEER + str(SEED_NUM)).upper()

print("Generated Access Key:", access_key)
```

```python
# [CELL 14] ATTEMPT CONTROL

attempt_limit = SEED_NUM + 3

print("Maximum Attempts:", attempt_limit)
```

```python
# [CELL 15] PASSWORD AUTHENTICATION

attempts = 0
access_granted = False

while True:
    entered_key = input("Enter Access Key: ").strip().upper()
    attempts += 1

    if entered_key == access_key:
        print("ACCESS GRANTED")
        access_granted = True
        break

    print("ACCESS DENIED")

    if attempts == attempt_limit:
        print("Maximum attempts reached.")
        break

    print("Attempts remaining:", attempt_limit - attempts)
```

```python
# [CELL 16] FINAL SYSTEM STATE

if access_granted:
    access_result = "ACCESS GRANTED"
    system_state = "UNLOCKED"
else:
    access_result = "ACCESS DENIED"
    system_state = "LOCKED"

print("Access Result:", access_result)
print("Final System State:", system_state)
print("Attempts Made:", attempts)
```
